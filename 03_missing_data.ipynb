```python
import numpy as np
import pandas as pd
```

```python
# data
# nan - not a number
data = {
    'A' : [1,2,np.nan,4,5],
    'B' : [1,2,3,4,5],
    'C' : [1,2,3,np.nan,np.nan],
    'D' : [1,np.nan,np.nan,np.nan,5]
}
df = pd.DataFrame(data)
df
```

```python
# how to find the missing data
# print the boolean indexing of the data where nan is present it show True
df.isna()
```

```python
# count the nan value
df.isna().sum()
```

```python
# remove the missing data , when the nan is there it removes 
df.dropna()
```

```python
# thresh = when you use it we need to provide one value like (thresh = n) here n = non null value 
# here n = 3 means whoever row has 3 non null value it will printed
df.dropna(thresh=3)
```

```python
# here n = 4
df.dropna(thresh=4)
```

```python
# fill the missing data where the value is nan we need to fill those value
df.fillna(0) # it fill the value with 0 where nan present
```

```python
# this block of code will fill the value with particular value like if A column have nan value i fill with 0
# same for B have nan value it fill with 100, for C = 300, for D = 400
values = {
    'A' : 0,
    'B' : 100,
    'C' : 300,
    'D' : 400
}
values
df.fillna(value=values)
```
