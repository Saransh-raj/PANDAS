```python
import numpy as np
import pandas as pd
```

```python
# merge the two data frames

# data frame 1 :-
employees = pd.DataFrame({
    'Employee_id' : [1,2,3,4,5],
    'name' : ['John','Bob','Alice','Hary','Sam'],
    'department' :['HR','IT','Finance','Analytics','CA']
})

# data frame 2 :
salaries = pd.DataFrame({
    'Employee_id' : [1,2,3,6,7],
    'salary' : [10000,20000,30000,40000,50000],
    'bonus' : [3000,2000,4000,1000,5000]
})
```

```python
employees
```

```python
salaries
```

```python
# merge both the dataframe
# Note :- merging will happen on the basis of something common like here Employee_id is common
pd.merge(employees,salaries,on='Employee_id',how='inner')

# on - means on the basis of employee_id 
# how = 'inner' - means join this by inner like the common of both data frames
```

```python
pd.merge(employees,salaries,on='Employee_id',how='outer')

# how = outer - means it join both the data frame without intersection like the union of the both data frame
```

```python
# how = left - means merge on the basis of employees
pd.merge(employees,salaries,on='Employee_id',how='left')
```

```python
# how = right - means merge on the basis of salaries
pd.merge(employees,salaries,on='Employee_id',how='right')
```

```python
# concatenate
df1 = pd.DataFrame({
    'A':['A0','A1','A2'],
    'B': ['B0','B1','B2'],
    'C': ['C0','C1','C2']
})

df2 = pd.DataFrame({
    'A':['A3','A4','A5'],
    'B': ['B3','B4','B5'],
    'C': ['C3','C4','C5']
})

# concate on the basis of column
pd.concat([df1,df2])
```

```python
# concate on the basis of rows
pd.concat([df1,df2],axis=1)
```

```python
# joining two data frame
df3 = pd.DataFrame({
    'name':['Alice','Bob','Charlie'],
},  
    index=[1,2,3]
)

df4 = pd.DataFrame({
    'score' : [85,90,75],
}, index=[2,3,4])

df3.join(df4)
```
