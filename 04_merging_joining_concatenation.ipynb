```python
import numpy as np
import pandas as pd
```

```python
# merge the two data frames

# data frame 1 :-
employees = pd.DataFrame({
    'Employee_id' : [1,2,3,4,5],
    'name' : ['John','Bob','Alice','Hary','Sam'],
    'department' :['HR','IT','Finance','Analytics','CA']
})

# data frame 2 :
salaries = pd.DataFrame({
    'Employee_id' : [1,2,3,6,7],
    'salary' : [10000,20000,30000,40000,50000],
    'bonus' : [3000,2000,4000,1000,5000]
})
```

```python
employees
```

```python
salaries
```

```python
# merge both the dataframe
# Note :- merging will happen on the basis of something common like here Employee_id is common
pd.merge(employees,salaries)
```
