```python
import pandas as pd
```

```python
data = {
    "Name" : ["Saransh","Humera","Abhi"],
    "Age" : [20,21,23],
    "Marks" : [78,100,65]
}

pd.DataFrame(data)
```

```python
# list into dataframe :- 
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]

df2 = pd.DataFrame(data_list)

# writing column name manually 
columns = ["Name","Age","City","Salary"]
df2 = pd.DataFrame(data_list,columns = columns)
df2
```

```python
# to access only Name and Age 
df2[["Name","Salary"]]
```

```python
# creating a new column
df2["Designation"] = ["Doctor","Eng.","Doctor","Eng."]
df2
```

```python
# to removing the column
df2.drop("Designation",axis=1)
```

```python
df2 # designation is still there 
```

```python
# for deleting the designation permanentaly
df2.drop("Designation",axis = 1,inplace=True)
df2 # now Designation is deleted permanantly
```

```python
# conditional selection 
# i only want to print only student whose age is greater than 30
df2[df2["Age"] > 30]
```

```python
# i only want to print only student whose age is greater than 30 and belongs from paris
df2[ (df2["Age"] > 30)  & (df2["City"] == "Paris")]
```
